```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("menu.csv")

# Display first rows
df.head()
```

```python
# Shape
print(df.shape)

# Columns
print(df.columns)

# Data types
df.info()

# Missing values
df.isnull().sum()
```

```python
# Check duplicates
df.duplicated().sum()

# Remove duplicates
df = df.drop_duplicates()

# Replace missing values
df.fillna(0, inplace=True)

# Convert columns if required
df.dtypes
```

```python
df.mean(numeric_only=True)
```

```python
df.median(numeric_only=True)
```

```python
df.mode()
df['Category'].value_counts()
```

```python
plt.figure(figsize=(10,5))

df['Category'].value_counts().plot(kind='bar')

plt.title("Items by Category")
plt.xlabel("Category")
plt.ylabel("Count")

plt.show()
```

```python
top_calories = df.sort_values(
    by="Calories",
    ascending=False
).head(10)

top_calories[['Item','Calories']]
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=top_calories,
    x="Calories",
    y="Item"
)

plt.title("Top 10 High Calorie Items")

plt.show()
```

```python
nutrition = df[
[
'Calories',
'Total Fat',
'Sodium',
'Carbohydrates',
'Sugars',
'Protein'
]
]

nutrition.mean()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df.corr(numeric_only=True),
    cmap="coolwarm"
)

plt.title("Nutrition Correlation Heatmap")

plt.show()
```

```python
df.sort_values(
    by="Protein",
    ascending=False
)[['Item','Protein']].head(10)
```

```python
protein=df.sort_values(
    by="Protein",
    ascending=False
).head(10)


plt.figure(figsize=(10,5))

sns.barplot(
    data=protein,
    x="Protein",
    y="Item"
)

plt.title("Top Protein Rich Items")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Calories'],
    bins=30
)

plt.title("Calories Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    y=df['Sugars']
)

plt.title("Sugar Level Analysis")

plt.show()
```
